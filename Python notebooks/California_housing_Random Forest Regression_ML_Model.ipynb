🧪 Practical Lesson 2 — Train/Test Split

🎯 Objective

We want to train a model that learns the relationship between:

Housing features → Median house value

But there's a problem:

If we train and evaluate the model on the same data, we can't properly know whether it can generalize to new houses.

So we divide the data into:

                 FULL DATASET
                      ↓
              Train/Test Split
                 ↙        ↘
          Training data   Test data
             ↓               ↓
          LEARN           UNSEEN
             ↓               ↓
          Model        Final evaluation

# Build the Preprocessing Pipeline
"Whenever new data arrives, perform these preprocessing steps in this exact order."

This prevents mistakes and makes the workflow reproducible.

```python
# Step 1 — Separate numerical and categorical columns
```

```python
categorical_cols = ["ocean_proximity"]
```

```python
# And we can automatically identify the numerical columns:
```

```python
numeric_cols = X_train.select_dtypes(include="number").columns.tolist()
```

```python
# Step 2 — Import the tools
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
# Step 3 — Build the numerical preprocessing pipeline
```

```python
numeric_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median")),("scaler", StandardScaler())])
```

```python
# Step 4 — Build the categorical preprocessing pipeline
```

```python
categorical_pipeline = Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                                 ("encoder", OneHotEncoder(handle_unknown="ignore"))]
                               )
```

```python
# Step 5 — Combine them with ColumnTransformer
```

```python
preprocessor = ColumnTransformer([("numerical", numeric_pipeline, numeric_cols),
                                  ("categorical", categorical_pipeline, categorical_cols)
                                 ])
```

```python
# Step 6 — Fit the preprocessor
```

```python
preprocessor.fit(X_train)
```

```python
X_train_processed = preprocessor.transform(X_train)
```

```python
X_test_processed = preprocessor.transform(X_test)
```

# 🌲 Model #3 — Random Forest Regression

🧠 Step 1 — Understand Random Forest

A Random Forest is essentially:

Many Decision Trees working together.

Instead of trusting one tree:

Data
 ↓
🌳 One Decision Tree
 ↓
Prediction

we build many trees:

                 Data
                  ↓
        ┌─────────┼─────────┐
        ↓         ↓         ↓
       🌳        🌳        🌳   ... many trees
        ↓         ↓         ↓
      $280k     $310k     $295k
        └─────────┼─────────┘
                  ↓
             Average
                  ↓
             💰 $295k

```python
# 🚀 Step 2 — Import
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
# Step 3 — Create Model #3
```

```python
model_3 = RandomForestRegressor(
    random_state=42
)
```

```python
# Step 4 — Train
```

```python
model_3.fit(X_train_processed, y_train)
```

```python
# Step 5 — Predict
```

```python
y_pred_3 = model_3.predict(X_test_processed)
```

# 📊 Next — Evaluate Model #3

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae_3 = mean_absolute_error(y_test, y_pred_3)
mse_3 = mean_squared_error(y_test, y_pred_3)
rmse_3 = np.sqrt(mse_3)
r2_3 = r2_score(y_test, y_pred_3)

print("Model #3 — Random Forest Regression")
print("MAE :", mae_3)
print("MSE :", mse_3)
print("RMSE:", rmse_3)
print("R²  :", r2_3)
```

# 🥇 Winner: Random Forest Regression

It wins on all three primary metrics:

Lowest MAE → predictions are closest on average.
Lowest RMSE → substantially fewer/severer large errors.
Highest R² → explains the largest proportion of variation.

So our decision is:

Random Forest Regression is the best candidate among the three models we tested. ✅

```python
import gradio as gr
import pandas as pd
```

```python
# Create the prediction function
```

```python
def predict_house_value(
    longitude,
    latitude,
    housing_median_age,
    total_rooms,
    total_bedrooms,
    population,
    households,
    median_income,
    ocean_proximity
):
    
    # Feature engineering
    rooms_per_household = total_rooms / households
    bedrooms_per_room = total_bedrooms / total_rooms
    population_per_household = population / households

    # Create raw input DataFrame
    new_house = pd.DataFrame({
        "longitude": [longitude],
        "latitude": [latitude],
        "housing_median_age": [housing_median_age],
        "total_rooms": [total_rooms],
        "total_bedrooms": [total_bedrooms],
        "population": [population],
        "households": [households],
        "median_income": [median_income],
        "rooms_per_household": [rooms_per_household],
        "bedrooms_per_room": [bedrooms_per_room],
        "population_per_household": [population_per_household],
        "ocean_proximity": [ocean_proximity]
    })

    # Prediction
    prediction = final_model.predict(new_house)[0]

    return f"${prediction:,.0f}"
```

# 🖥️ Step  — Build the interface

```python
interface = gr.Interface(
    fn=predict_house_value,
    
    inputs=[
        gr.Number(label="Longitude"),
        gr.Number(label="Latitude"),
        gr.Number(label="Housing Median Age"),
        gr.Number(label="Total Rooms"),
        gr.Number(label="Total Bedrooms"),
        gr.Number(label="Population"),
        gr.Number(label="Households"),
        gr.Number(label="Median Income"),
        gr.Dropdown(
            choices=[
                "<1H OCEAN",
                "INLAND",
                "ISLAND",
                "NEAR BAY",
                "NEAR OCEAN"
            ],
            label="Ocean Proximity"
        )
    ],
    
    outputs=gr.Textbox(label="Predicted Median House Value"),
    
    title="🏠 California House Value Predictor",
    
    description="Enter house features and let our Random Forest model estimate the median house value."
)
```

```python
import gradio as gr

with gr.Blocks(title="California House Value Predictor") as app:

    gr.Markdown(
        """
        # 🏠 California House Value Predictor
        ### AI-powered median house-value estimation

        Enter the characteristics of a California housing district
        and let our trained **Random Forest** model estimate its
        median house value.
        """
    )

    with gr.Row():

        with gr.Column():
            gr.Markdown("### 📍 Location")

            longitude = gr.Number(
                label="Longitude",
                value=-118.25
            )

            latitude = gr.Number(
                label="Latitude",
                value=34.05
            )

        with gr.Column():
            gr.Markdown("### 🏡 Property")

            housing_median_age = gr.Number(
                label="Housing Median Age",
                value=25
            )

            total_rooms = gr.Number(
                label="Total Rooms",
                value=2500
            )

            total_bedrooms = gr.Number(
                label="Total Bedrooms",
                value=500
            )

    with gr.Row():

        with gr.Column():
            gr.Markdown("### 👨‍👩‍👧‍👦 Population & Households")

            population = gr.Number(
                label="Population",
                value=1200
            )

            households = gr.Number(
                label="Households",
                value=450
            )

        with gr.Column():
            gr.Markdown("### 💵 Income & Location Type")

            median_income = gr.Number(
                label="Median Income",
                value=6.5
            )

            ocean_proximity = gr.Dropdown(
                choices=[
                    "<1H OCEAN",
                    "INLAND",
                    "ISLAND",
                    "NEAR BAY",
                    "NEAR OCEAN"
                ],
                value="NEAR OCEAN",
                label="Ocean Proximity"
            )

    predict_button = gr.Button(
        "🔮 Predict House Value",
        variant="primary"
    )

    prediction_output = gr.Markdown(
        "### 💰 Prediction will appear here"
    )

    gr.Markdown(
        """
        ---
        **Model:** Random Forest Regression  
        **R²:** 0.8067  

        *The prediction is an ML estimate based on patterns learned
        from the California Housing dataset. It is not a guaranteed
        market valuation.*
        """
    )

    # 🔌 CONNECT BUTTON TO OUR ML MODEL
    predict_button.click(
        fn=predict_house_value,
        inputs=[
            longitude,
            latitude,
            housing_median_age,
            total_rooms,
            total_bedrooms,
            population,
            households,
            median_income,
            ocean_proximity
        ],
        outputs=prediction_output
    )


# 🚀 Launch the application
app.launch()
```
