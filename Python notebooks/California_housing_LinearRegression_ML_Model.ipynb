🧪 Practical Lesson 2 — Train/Test Split

🎯 Objective

We want to train a model that learns the relationship between:

Housing features → Median house value

But there's a problem:

If we train and evaluate the model on the same data, we can't properly know whether it can generalize to new houses.

So we divide the data into:

                 FULL DATASET
                      ↓
              Train/Test Split
                 ↙        ↘
          Training data   Test data
             ↓               ↓
          LEARN           UNSEEN
             ↓               ↓
          Model        Final evaluation

```python
numeric_cols = X_train.select_dtypes(include="number").columns

outlier_summary = []

for col in numeric_cols:
    Q1 = X_train[col].quantile(0.25)
    Q3 = X_train[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((X_train[col] < lower) | (X_train[col] > upper)).sum()

    outlier_summary.append([col, count, lower, upper])

outlier_summary = pd.DataFrame(
    outlier_summary,
    columns=["Feature", "Outlier_Count", "Lower_Bound", "Upper_Bound"]
)

outlier_summary
```

# Build the Preprocessing Pipeline
"Whenever new data arrives, perform these preprocessing steps in this exact order."

This prevents mistakes and makes the workflow reproducible.

```python
# Step 1 — Separate numerical and categorical columns
```

```python
categorical_cols = ["ocean_proximity"]
```

```python
# And we can automatically identify the numerical columns:
```

```python
numeric_cols = X_train.select_dtypes(include="number").columns.tolist()
```

```python
# Step 2 — Import the tools
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
# Step 3 — Build the numerical preprocessing pipeline
```

```python
numeric_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median")),("scaler", StandardScaler())])
```

```python
# Step 4 — Build the categorical preprocessing pipeline
```

```python
categorical_pipeline = Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                                 ("encoder", OneHotEncoder(handle_unknown="ignore"))]
                               )
```

```python
# Step 5 — Combine them with ColumnTransformer
```

```python
preprocessor = ColumnTransformer([("numerical", numeric_pipeline, numeric_cols),
                                  ("categorical", categorical_pipeline, categorical_cols)
                                 ])
```

```python
# Step 6 — Fit the preprocessor
```

```python
preprocessor.fit(X_train)
```

```python
X_train_processed = preprocessor.transform(X_train)
```

```python
X_test_processed = preprocessor.transform(X_test)
```

```python
# Step 7 Check the result
```

```python
X_train_processed.shape
X_test_processed.shape
```

```python
print("Training data:", X_train_processed.shape)
print("Test data:", X_test_processed.shape)
```

# Train Our First Linear Regression Model

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train_processed, y_train)

y_pred = model.predict(X_test_processed)
```

```python
print("First 10 predictions:")
print(y_pred[:10])
```

```python
print("\nFirst 10 actual values:")
print(y_test.iloc[:10].values)
```

MAE

come in.

# Lesson 6 — Model Evaluation

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)
```

```python
# Next:MSE — Mean Squared Error
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)

print("MSE:", mse)
```

# RMSE

```python
from sklearn.metrics import root_mean_squared_error

rmse = root_mean_squared_error(y_test, y_pred)

print("RMSE:", rmse)
```

# 🎯 Now the final major metric: R²

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R²:", r2)
```

# ✅ Approximately 59.7% of the variance in median_house_value is explained by the model on the test set.
